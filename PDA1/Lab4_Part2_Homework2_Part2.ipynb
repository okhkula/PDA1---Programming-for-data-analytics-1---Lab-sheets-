**5.9 (Palindrome Tester)** A string that spelled identically backward and forward, like 'radar' is a palindrome. Write a function is_palindrome that takes a string and returns True if itâ€™s a palindrome and False otherwise. Use a stack (simulated with a list as we did in Section 5.11) to help determine whether a string is a palindrome. Your function should ignore case sensitivity (that is, 'a' and 'A' are the same), spaces and punctuation.

```python

import string

def is_pelindrome(s):
  s=s.lower()

  stack=[]

  for char in s:
    stack.append(char)

  for char in s:
    if char!=stack.pop():
      return False
  return True

print(is_pelindrome("radar"))
```

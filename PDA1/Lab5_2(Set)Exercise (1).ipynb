Sets:




*   Sets are collections of unique elements.
*   They are unordered, meaning the elements do not have a specific order
*   Created using curly braces {} or the set() function.
*   Useful for operations like finding intersections, unions, differences between sets.

Example:

*   my_set = {1, 2, 3, 4}

Example2:
* my_list = [1, 2, 3, 4, 5]
* my_set = set(my_list)
* print(my_set)  # Output: {1, 2, 3, 4, 5}


### **creating sets, checking membership, and basic iteration**

```python

# Creating a set with colors
my_set = {"red", "blue", "green", "yellow", "red"}  # Including a duplicate "red"
print("Original set:", my_set)
```

```python
#Checking if a color is in the set
'blue' in my_set
```

```python
'purple' in my_set
```

```python
# Iterating through the set
print("\nIterating through the set:")
for color in my_set:
    print(color)
```

```python
# Adding elements to the set
my_set.add("orange")  # Adding a new color
print("\nAfter adding 'orange' to the set:", my_set)
```

```python
#remove elements to the set
my_set.remove("orange")  # Adding a new color
print("\nAfter adding 'orange' to the set:", my_set)
```

```python
my_set.pop()
```

```python
my_set
```

```python
# Discarding elements from the set (won't raise error if element doesn't exist)
my_set.discard("purple")  # Discarding a color that doesn't exist
print("After discarding 'purple' (which doesn't exist) from the set:", my_set)
```

```python
# Discarding elements from the set (won't raise error if element doesn't exist)
my_set.discard("red")  # Discarding a color that doesn't exist
print("After discarding 'purple' (which doesn't exist) from the set:", my_set)
```

```python
# Clearing the set
my_set.clear()
print("After clearing the set:", my_set)
```

## **comparing sets, mathematical set operations, mutable set operators and methods, and set comprehensions.**

```python
# Importing necessary libraries
import random

# Generating random data for sets
set1 = set(random.sample(range(1, 100), 20))
set2 = set(random.sample(range(1, 100), 20))
```

```python
# Displaying the sets
print("Set 1:", set1)
print("Set 2:", set2)
```

**Comparing sets**

```python
# subset

print("Set 1 is subset of Set 2:", set1.issubset(set2))
```

```python
#superset
print("Set 2 is superset of Set 1:", set2.issuperset(set1))
```

```python
#intersection
print("Set 1 and Set 2 have common elements:", set1.intersection(set2))
```

```python
# Using the & operator for intersection
common_elements = set1 & set2
print("Set 1 and Set 2 have common elements:", common_elements)

```

```python
#Union

print("Union of Set 1 and Set 2:", set1.union(set2))
```

```python
Union = set1 | set2
print("Union of Set 1 and Set 2", Union)
```

```python
#Difference
print("Difference of Set 1 from Set 2:", set1.difference(set2))
```

```python
Difference = set1 - set2
print("Difference of Set 1 and Set 2", Difference)
```

```python
#Symmetric Difference
print("Symmetric Difference of Set 1 and Set 2:", set1.symmetric_difference(set2))
```

```python
Symmetric_Difference = set1 ^ set2
print("Difference of Set 1 and Set 2", Symmetric_Difference)
```

# Splitting text into words, creating a set of unique words, sorting them, and printing each word in sorted order


```python
# Define the text
text = 'to be or not to be that is the question'

words= text.split()

print("Words are",words)

```

```python
# Split the text into words and create a set of unique words
unique_words = set(words)

# Iterate over the sorted unique words and print them
for word in sorted(unique_words):
    print(word, end=' ')
```

# "Example 1: Analyze customer purchases and loyalty program membership to gain insights. Use the provided datasets to answer questions such as:**

* What are the most frequently purchased products?
* How many customers are part of the loyalty program?
* Are there any differences in purchasing behavior between loyalty program members and non-members?
*Perform the necessary analysis to obtain these results."

```python
import pandas as pd

# Sample data for purchases_2023.csv
purchases_data = {
    'customer_id': [101, 102, 103, 104, 105],
    'product_id': [1001, 1002, 1003, 1004, 1005],
    'quantity': [2, 1, 3, 1, 2],
    'price': [20.50, 10.99, 15.75, 8.00, 30.25]
}

purchases_df = pd.DataFrame(purchases_data)

# Sample data for loyalty_members.csv
loyalty_data = {
    'customer_id': [101, 103, 105]
}

loyalty_df = pd.DataFrame(loyalty_data)


```

```python
# Convert customer IDs to sets
all_customers = set(purchases_df['customer_id'])
loyalty_customers = set(loyalty_df['customer_id'])

# Determine the total number of unique customers who made purchases in 2023
total_customers = len(all_customers)

# Find out how many customers are loyalty program members
loyalty_members_count = len(loyalty_customers)

# Identify the set of customers who made purchases but are not part of the loyalty program
non_loyalty_customers = all_customers - loyalty_customers

# Display results
print("Total number of unique customers who made purchases in 2023:", total_customers)
print("Number of customers who are part of the loyalty program:", loyalty_members_count)
print("Number of customers who made purchases but are not part of the loyalty program:", len(non_loyalty_customers))
```

#**Exercise:**

* Identify the total number of unique product categories sold on the platform.
* Determine the number of customers who purchased products from at least three different categories.
* Find out the set of customers who purchased products from the "Electronics" category but did not purchase products from the "Clothing" category.

```python
import pandas as pd

# Sample data for customer_purchases.csv
customer_purchases_data = {
    'customer_id': [101, 102, 103, 101, 104, 102, 105, 103, 106, 107, 108,101,101,101],
    'product_id': [1001, 1002, 1003, 1004, 1005, 1001, 1003, 1002, 1004, 1005, 1001,1002,1003,1004]
}
```

```python
customer_purchases_df = pd.DataFrame(customer_purchases_data)
customer_purchases_df
```

```python

# Sample data for product_categories.csv
product_categories_data = {
    'product_id': [1001, 1002, 1003, 1004, 1005],
    'category': ['Electronics', 'Clothing', 'Electronics', 'Clothing', 'Accessories']
}

```

```python
product_categories_df = pd.DataFrame(product_categories_data)
product_categories_df
```

```python
# Extract unique product categories
total_categories = set(product_categories_data['category'])
print(total_categories)
```

```python
# Extract unique customer IDs who purchased from at least three different categories
customer_categories = {}

for _, row in customer_purchases_df.iterrows():
    customer_categories.setdefault(row['customer_id'], set()).add(
        product_categories_df.loc[product_categories_df['product_id'] == row['product_id'], 'category'].values[0])

customers_three_plus = [customer for customer, categories in customer_categories.items() if len(categories) >= 3]
print(customers_three_plus)
```

```python
# Extract customers who purchased from "Electronics" but not from "Clothing"
electronics_buyers = {customer for customer, categories in customer_categories.items() if 'Electronics' in categories}
clothing_buyers = {customer for customer, categories in customer_categories.items() if 'Clothing' in categories}
print(electronics_buyers - clothing_buyers)
```

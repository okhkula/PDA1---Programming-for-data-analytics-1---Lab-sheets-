# Exercise 4.9
(Temperature Conversion) Implement a fahrenheit function that returns the
Fahrenheit equivalent of a Celsius temperature. Use the following formula:

> F = (9 / 5) * C + 32

Use this function to print a chart showing the Fahrenheit equivalents of all Celsius temperatures
in the range 0–100 degrees. Use one digit of precision for the results. Print the
outputs in a neat tabular format.

```python
# Exercise 4.9
import pandas as pd

dp = pd.DataFrame(columns=['Celsius', 'Fahrenheit'])

def fahrenheit(C):
    return round((9/5) *C +32, 1)

data=[]
for C in range(0, 101):
    data.append({'Celcsious':C, 'Fahrenheit':fahrenheit(C)})
dp=pd.DataFrame(data)
print("\npandas data Frame:\n",dp)
```

number to be guessed by selecting a random integer in the range 1 to 1000. Do not reveal

the correct answer, display "Congratulations. You guessed the number!", and allow the
user to choose whether to play again.

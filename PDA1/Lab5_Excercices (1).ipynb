***(Counting Duplicate Words)*** Write a script that uses a dictionary to determine and print the number of duplicate words in a sentence. Treat uppercase and lowercase letters the same and assume there is no punctuation in the sentence. Use the techniques you learned in Lecture. Words with counts larger than 1 have duplicates.

````python
Exampletext = ('this is sample text with several words '
        'this is more sample text with some different words')


````

````python
from collections import Counter
text = ('this is sample text with several words '
        'this is more sample text with some different words')
counter = Counter(text.split())
for word, count in sorted(counter.items()):
    print(f'{word:<12}{count}')
````

**3. _(IPython Session)_** Read the file `grades.txt` that you created in the previous section and display it in columns with the column heads `'ID'`, `'Name'` and `'Grade'`.

**Answer:**

````python
with open('grades.txt', 'r') as grades:
    print(f'{"ID":<4}{"Name":<7}{"Grade"}')
    for record in grades:
        student_id, name, grade = record.split()
        print(f'{student_id:<4}{name:<7}{grade}')
````

**(Class Average: Writing Student Records to a CSV File)** An instructor teaches a
class in which each student takes three exams. The instructor would like to store this information
in a file named grades.csv for later use. Write code that enables an instructor
to enter each student’s first name and last name as strings and the student’s three exam
grades as integers. Use the csv module to write each record into the grades.csv file. Each
record should be a single line of text in the following CSV format:

*firstname,lastname,exam1grade,exam2grade,exam3grade*

````python
import csv
````

````python
with open('grades.csv', mode='w') as grades:
    writer = csv.writer(grades)
    firstname=input('Enter first name: ')
    lastname=input('Enter last name: ')
    exam1grade=int(input('Enter exam 1 grade: '))
    exam2grade=int(input('Enter exam 2 grade: '))
    exam3grade=int(input('Enter exam 3 grade: '))
    f=([firstname,lastname,exam1grade,exam2grade,exam3grade])
    writer.writerow(f)
    print(f)
````

**(Class Average: Reading a Gradebook Dictionary from a JSON File)** use the json module to read the grades.json file created in the previous
exercise. Display the data in tabular format, including an additional column showing each
student’s average to the right of that student’s three exam grades and an additional row
showing the class average on each exam below that exam’s column. (Hint: you may use Pandas json_normalize() function if you want)

documentation:
https://pandas.pydata.org/docs/reference/api/pandas.json_normalize.html

````python
import pandas as pd
import json


with open('grades.json', 'r') as f:
    grades_data = json.load(f)


mydata_df = pd.json_normalize(grades_data['students'])

av_list = []
av_list.append(mydata_df['exam1 '].mean())
av_list.append(mydata_df['exam2 '].mean())
av_list.append(mydata_df['exam3 '].mean())
av_list.append(mydata_df['average'].mean())
print(av_list)


mydata_df.loc[0] = ['---', '----', av_list[0], av_list[1], av_list[2], av_list[3]]

display(mydata_df)
````

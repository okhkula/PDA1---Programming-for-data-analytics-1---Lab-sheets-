***Challenge: Writing the Word Equivalent of a Check Amount***

In check-writing systems, it’s crucial to prevent alteration of check amounts. One common security method requires that the amount be written in numbers and spelled out in words as well. Even if someone can alter the numerical amount of the check, it’s tough to change the amount in words. Create a dictionary that maps numbers to their corresponding word equivalents.

Write a script that inputs a numeric check amount that’s less than 1000 and uses the dictionary to write the word equivalent of the amount. For example, the amount 112.43 should be written as ONE HUNDRED TWELVE AND 43/100

```python
number_to_word = {
    0: 'ZERO', 1: 'ONE', 2: 'TWO', 3: 'THREE', 4: 'FOUR',
    5: 'FIVE', 6: 'SIX', 7: 'SEVEN', 8: 'EIGHT', 9: 'NINE',
    10: 'TEN', 11: 'ELEVEN', 12: 'TWELVE', 13: 'THIRTEEN',
    14: 'FOURTEEN', 15: 'FIFTEEN', 16: 'SIXTEEN', 17: 'SEVENTEEN',
    18: 'EIGHTEEN', 19: 'NINETEEN', 20: 'TWENTY', 30: 'THIRTY',
    40: 'FORTY', 50: 'FIFTY', 60: 'SIXTY', 70: 'SEVENTY',
    80: 'EIGHTY', 90: 'NINETY'
}

#Use the above dictionary to implement the functionality described in the question
def convert_to_word(num):
  word=[]
  if num >=100:
    hundreds=num//100
    word.append(number_to_word[hundreds]+" HUNDRED")
    num%=100

    if num>=20:
      tens=(num//10)*10
      word.append(number_to_word[tens])
      num%=10

    if num>0:
      word.append(number_to_word[num])

    return " ".join(word)

amount=float(input("Enter the amount:"))
dollars=int(amount)
cents=round((amount-dollars)*100)

dollars_in_word=convert_to_word(dollars)


if cents>0:
  cents_in_words=f'{cents}/100'
else:
  cents_in_words=''


if dollars_in_word and cents_in_words:
  print(f'{dollars_in_word} AND {cents_in_words}')
elif dollars_in_word:
  print(dollars_in_word)
elif cents_in_words:
  print(cents_in_words)
else:
  print('ZERO')


```

Question 1: (old Latin) Write a script that encodes English-language phrases into a form of
coded language called old Latin. There are many different ways to form old Latin phrases.

For simplicity, use the following algorithm:
To form a old Latin phrase from an English-language phrase, tokenize the phrase
into words with string method split. To translate each English word into a old Latin word, place the first letter of the English word at the end of the word and add the letters “ay.” Thus, the word “jump” becomes “umpjay,” the word “the” becomes “hetay,” and the word “computer” becomes “omputercay.” If the word starts with a vowel, just add “ay.”
Blanks between words remain as blanks. Assume the following: The English phrase consists of words separated by blanks, there are no punctuation marks and all words have two or more letters. Enable the user to enter a sentence, then display the sentence in big Latin.

```python
def old_latin(sentence):
    words=sentence.split()
    vowels=['a','e','i','o','u']
    old_latin_words=''
    for word in words:
        if word[0].lower() in vowels:
            old_latin_words=word+'ay'
        else:
            old_latin_words=(word[1:]+word[0]+'ay')
    return old_latin_words.strip()


sentence=input("Enter a sentence: ")
latin_translate=old_latin(sentence)
print(latin_translate)
```

Question 2: (Reversing a Sentence) Write a script that reads a line of text as a string, tokenizes the string with the **split** method and outputs the tokens in reverse order. Use space characters as delimiters.

```python
sentence=input("Enter a sentence: ")
tokens=sentence.split()
tokens.reverse()
''.join(tokens)
print(tokens)
```

Question 3: (Tokenizing and Comparing Strings) Write a script that reads a line of text, tokenizes
the line using space characters as delimiters and outputs only those words beginning
with the letter 'b'.

```python
text=input("Enter a text: ")
tokens=text.split()
for word in tokens:
    if word.startswith('b'):
        ''.join(word)
        print(word)
```

Question 4: (Tokenizing and Comparing Strings) Write a script that reads a line of text, tokenizes
it using space characters as delimiters and outputs only those words ending with
the letters 'ed'.

```python
text=input("Enter a text:")
words=text.split()
for word in words:
  word.lower()
  if word.endswith('ed'):
    ''.join(word)
    print(word)
```

Question 5: ( Simple Sentiment Analysis) Search online for lists of positive sentiment
words and negative sentiment words. Create a script that inputs text, then determines
whether that text is positive or negative, based on the total number of positive words and
the total number of negative words. Test your script by searching for Twitter tweets on a
topic of your choosing, then entering the text for several tweets. In the data science case
study chapters, we’ll take a deeper look at sentiment analysis.

```python
postive_words=["amazing", "bright", "creative", "delightful", "energetic",
    "friendly", "gracious", "happy", "inspiring", "joyful",
    "kind", "loving", "motivated", "optimistic", "powerful",
    "radiant", "strong", "thriving", "uplifting", "vibrant"]

negative_words=["angry", "anxious", "awful", "bad", "bitter",
    "boring", "broken", "chaotic", "confused", "cruel",
    "depressed", "difficult", "disappointed", "disgusted", "disturbed",
    "embarrassed", "fearful", "frustrated", "gloomy", "greedy",
    "guilty", "hateful", "helpless", "hopeless", "hostile",
    "hurt", "ignorant", "impatient", "insecure", "jealous"]

sentence=input("Enter a sentence: ")
words=sentence.split()
postive_count=0
negative_count=0
for word in words:
  if word.lower() in postive_words:
    postive_count+=1
  elif word.lower() in negative_words:
    negative_count+=1
if postive_count>negative_count:
  print("positive")
elif postive_count==negative_count:
  print("neutral")
else:
  print("negative")
```
